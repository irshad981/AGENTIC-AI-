## LCEL and the Runnable Interface ##

```python
# Stream the response for real-time output
print("🔄 Streaming response:")
print("-" * 50)

stream = llm_client.stream([
    SystemMessage(content="You are a product developer"),
    HumanMessage(content="tell me about the NPCI PRODUCT AEPS AND FASTAG")
])

for chunk in stream:
    print(chunk.content, end="", flush=True)

print("\n" + "-" * 50)
print("✅ Streaming complete!")
```

## BATCH PROCESSING ###

```python
# Prepare multiple inputs for batch processing
batch_inputs = [
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
]

# Process in batch
batch_results = llm_client.batch(batch_inputs, config={
        'max_concurrency': 5
    })

print("📦 Batch Processing Results:")
print("=" * 60)
for i, result in enumerate(batch_results, 1):
    print(f"\n{i}. {result.content}...")
    print("-" * 40)
```

## JSON Output Parser ##


```python
from langchain_core.output_parsers import StrOutputParser, JsonOutputParser

# Create a JSON parser
parser = JsonOutputParser()

# Get a JSON-style response from the model
response = llm_client.invoke([
    HumanMessage(content="""Tell me about elon musk, history, education, 
                 work experience, skills and achievements. Give the response in json format.""")
])

parsed_output = parser.invoke(response)

print("📝 Raw Output:" , response.content)
print(f"Raw response type: {type(response)}")
print(f"Parsed output type: {type(parsed_output)}")
print(f"Parsed content: {parsed_output}")
```

##  Simple Chain – LLM + JSON Parser ##


```python
# Create a simple chain: LLM + JSON Parser
simple_chain = llm_client | JsonOutputParser()

# Test the chain
result = simple_chain.invoke([
    HumanMessage(content="""Tell me about elon musk, history, education, 
                 work experience, skills and achievements. Give the response in json format.""")
])

print(f"Chain result type: {type(result)}")
print(f"Chain result: {result}")
```

##  Prompt Templates in LangChain 1.0 ##

```python
from langchain_core.prompts import ChatPromptTemplate

# Create a prompt template
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are an expert in {subject}. Provide clear, concise explanations."),
    ("human", "Explain {topic} in simple terms.")
])

# Format the prompt
formatted_prompt = prompt_template.invoke({
    "subject": "software development", 
    "topic": "the differences between LangChain 0.x and 1.0"
})

print("📋 Formatted Prompt:")
for message in formatted_prompt.messages:
    print(f"{message.__class__.__name__}: {message.content}")
```

Complete Chain – Prompt + LLM + String Parser

```python
# Create a complete chain
complete_chain = prompt_template | llm_client | StrOutputParser()

# Use the chain
explanation = complete_chain.invoke({
    "subject": "AI frameworks",
    "topic": "LangChain 1.0's unified interface design"
})

print("Complete Chain Result:")
print(explanation)
```

## Structured Data with Pydantic Output Parser ##


```python
from typing import List
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field


# Define data models using Pydantic
class Person(BaseModel):
    """Information about a person."""
    name: str = Field(description="The name of the person")
    height_in_meters: float = Field(description="The height of the person expressed in meters.")

class People(BaseModel):
    """Identifying information about all people in a text."""
    people: List[Person]

# Set up a Pydantic parser
pydantic_parser = PydanticOutputParser(pydantic_object=People)

print("📋 Pydantic Parser Format Instructions:")
print(pydantic_parser.get_format_instructions())
```

## Using Pydantic Parser with ChatPromptTemplate ##


```python
# Create prompt with format instructions
structured_prompt = ChatPromptTemplate.from_messages([
    ("system", "Answer the user query. Wrap the output in `json` tags\n{format_instructions}"),
    ("human", "{query}")
]).partial(format_instructions=pydantic_parser.get_format_instructions())

# Create chain with Pydantic parser
structured_chain = structured_prompt | llm_client 

# Test the structured output
query = "Siva is 44 years old and 1.7 meters tall. Rishik is 15 years old and 1.8 meters tall."
result = structured_chain.invoke({"query": query})

print("🔗 Structured Chain Result:")
print(f"Type: {type(result).__name__}")
print(f"Content: {result.content}")


parsed_people = pydantic_parser.invoke(result)
print(f"\n✅ Parsed People: {parsed_people}")
print(f"People type: {type(parsed_people)}")
print(f"First person: {parsed_people.people[0].name} - {parsed_people.people[0].height_in_meters}m")
```
