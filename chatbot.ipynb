```python

import gradio as gr

def predict(message, history):
    if not message.strip():
        return history, history

    print("Message:", message)
    print("History:", history)

    # Keep history as a list of user/assistant dictionaries
    messages = list(history or [])

    messages.append({
        "role": "user",
        "content": message
    })

    response = get_completion_from_messages(messages)

    messages.append({
        "role": "assistant",
        "content": response
    })

    # Convert messages into tuples for the older Chatbot component
    chat_history = []
    for i in range(0, len(messages), 2):
        if i + 1 < len(messages):
            chat_history.append((
                messages[i]["content"],
                messages[i + 1]["content"]
            ))

    return chat_history, messages


with gr.Blocks() as demo:
    gr.Markdown("# 🤖 Chatbot Interface")
    gr.Markdown("Type your message and get a response from the assistant.")

    chatbot = gr.Chatbot(elem_id="chatbot")

    msg = gr.Textbox(
        placeholder="Type your message here...",
        label="Your Message"
    )

    clear = gr.Button("Clear Chat")
    state = gr.State([])

    def user_submit(message, history):
        chat_history, updated_history = predict(message, history)
        return chat_history, updated_history, ""

    msg.submit(
        user_submit,
        inputs=[msg, state],
        outputs=[chatbot, state, msg]
    )

    clear.click(
        lambda: ([], []),
        outputs=[chatbot, state]
    )

demo.launch(debug=True)

```
